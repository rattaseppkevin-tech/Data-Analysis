```python
import pandas as pd
import numpy as np

pd.set_option('display.float_format', lambda x: '%.2f' % x)

sales = pd.read_csv(r"C:\Users\Kevin\Desktop\project_transactions.csv")
products = pd.read_csv(r"C:\Users\Kevin\Desktop\product.csv")
```

```python
sales["DAY"] = sales["DAY"].astype("int8")
sales["QUANTITY"] = sales["QUANTITY"].astype("int32")
sales["STORE_ID"] = sales["STORE_ID"].astype("int32")
sales["WEEK_NO"] = sales["WEEK_NO"].astype("uint8")
```

```python
sales.isna().sum()
```

```python
sales[["household_key", "PRODUCT_ID"]].nunique()
```

```python
sales["TOTAL_DISCOUNT"] = sales["RETAIL_DISC"] + sales["COUPON_DISC"]
sales["DISCOUNT_PERCENT"] = (sales["TOTAL_DISCOUNT"] / sales["SALES_VALUE"]).abs()
sales["DISCOUNT_PERCENT"] = sales["DISCOUNT_PERCENT"].clip(0, 1)
sales[["RETAIL_DISC", "COUPON_DISC", "TOTAL_DISCOUNT", "SALES_VALUE", "DISCOUNT_PERCENT"]].head()
```

```python
sales = sales.drop(["RETAIL_DISC", "COUPON_DISC", "COUPON_MATCH_DISC"], axis=1)
```

```python
sales.head()
```

```python
sales[["SALES_VALUE", "TOTAL_DISCOUNT", "QUANTITY"]].sum()
```

```python
overall_pct = (sales["TOTAL_DISCOUNT"].sum() / sales["SALES_VALUE"].sum())
overall_pct = abs(overall_pct)
overall_pct.round(2)
```

```python
max_qty = sales["QUANTITY"].max()

max_qty_row = sales.loc[sales["QUANTITY"] == max_qty]

print(f"Suurim kogus: {max_qty}")
max_qty_row
```

```python
household_sales = sales.groupby("household_key")["SALES_VALUE"].sum()
household_sales.hist()
```

```python
top_households = sales.groupby("household_key").agg({"QUANTITY": "sum", "SALES_VALUE": "sum"})
top_ten = top_households.sort_values(by="QUANTITY", ascending=False).head(10)
top_ten
```

```python
top_10_sales = (sales
                .groupby("household_key")["SALES_VALUE"]
                .sum()
                .sort_values(ascending=False)
                .head(10))

top_10_sales.plot(
    kind="bar",
    title="Top 10",
    ylabel= "Euro",
    xlabel= "ID")
```

```python
product_sales = sales.groupby("PRODUCT_ID")["SALES_VALUE"].sum()
top_10_product = product_sales.sort_values(ascending=False).head(10)
top_10_product.plot(kind="barh")
```

```python
sales_with_names = pd.merge(sales, products, on="PRODUCT_ID", how="left")
```

```python
top_10_names = (
    sales_with_names.groupby("COMMODITY_DESC")["SALES_VALUE"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# Joonista horisontaalne graafik (barh)
top_10_names.plot(kind="barh", figsize=(10, 6), title="Top 10 tootekategooriat käibe järgi")
```

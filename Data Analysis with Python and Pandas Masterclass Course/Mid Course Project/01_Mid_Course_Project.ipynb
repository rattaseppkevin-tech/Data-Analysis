```python
import pandas as pd
import numpy as np

pd.set_option('display.float_format', lambda x: '%.2f' % x)

sales = pd.read_csv(r"C:\Users\Kevin\Desktop\project_transactions.csv")
```

```python
sales["DAY"] = sales["DAY"].astype("int8")
sales["QUANTITY"] = sales["QUANTITY"].astype("int32")
sales["STORE_ID"] = sales["STORE_ID"].astype("int32")
sales["WEEK_NO"] = sales["WEEK_NO"].astype("uint8")
```

```python
sales.isna().sum()
```

```python
sales[["household_key", "PRODUCT_ID"]].nunique()
```

```python
sales["TOTAL_DISCOUNT"] = sales["RETAIL_DISC"] + sales["COUPON_DISC"]
sales["DISCOUNT_PERCENT"] = (sales["TOTAL_DISCOUNT"] / sales["SALES_VALUE"]).abs()
sales["DISCOUNT_PERCENT"] = sales["DISCOUNT_PERCENT"].clip(0, 1)
sales[["RETAIL_DISC", "COUPON_DISC", "TOTAL_DISCOUNT", "SALES_VALUE", "DISCOUNT_PERCENT"]].head()
```

```python
sales = sales.drop(["RETAIL_DISC", "COUPON_DISC", "COUPON_MATCH_DISC"], axis=1)
```

```python
sales.head()
```

```python
sales[["SALES_VALUE", "TOTAL_DISCOUNT", "QUANTITY"]].sum()
```

```python
overall_pct = (sales["TOTAL_DISCOUNT"].sum() / sales["SALES_VALUE"].sum())
overall_pct = abs(overall_pct)
overall_pct.round(2)
```

```python
import pandas as pd
```

```python
transactions = pd.read_csv(r"C:\Users\Kevin\Desktop\transactions.csv", parse_dates=["date"])
```

```python
transactions.tail()
```

```python
max_date = transactions["date"].max()

transactions = transactions.assign(
    year = transactions['date'].dt.year,
    month = transactions['date'].dt.month,
    dayofweek = transactions['date'].dt.dayofweek,
    days_since_max = max_date - transactions["date"]
)
```

```python
transactions = transactions.assign(time_to_last_date = transactions["days_since_max"] + (pd.to_timedelta(21, unit="D"))).drop(columns=["days_since_max"])
transactions.head()
```

```python
transactions = transactions.assign(weeks_to_last_date = transactions["time_to_last_date"] / (pd.to_timedelta(7, unit="D")))
transactions.head()
```

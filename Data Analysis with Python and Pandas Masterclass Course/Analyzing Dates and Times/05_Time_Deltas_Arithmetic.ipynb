```python
import pandas as pd
```

```python
retail = pd.read_csv(r"C:\Users\Kevin\Desktop\retail.csv", parse_dates=["date"])
```

```python
retail.info(memory_usage="deep")
```

```python
retail = retail.assign(birthday=pd.to_datetime("1980-02-02"))
retail.head()
```

```python
retail = retail.assign(age= (retail["date"] - retail["birthday"]).dt.days / 365)
retail.head()
```

```python
retail = retail.assign(retirement_date = retail["date"] + (pd.to_timedelta(366, unit="D")))
retail.head()
```
